# MLForge — Day 1 Data Understanding

Minimal inspection supporting the production data contract. No model training or feature fitting is performed here.

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
assert DATA_PATH.exists(), f"Dataset not found: {DATA_PATH}"
df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
df.head()
```

```python
pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing": df.isna().sum(),
    "unique": df.nunique(dropna=False),
})
```

```python
print(df["Churn"].value_counts(dropna=False))
print(df["Churn"].value_counts(normalize=True, dropna=False).mul(100).round(2))
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customerID"].duplicated().sum())
```

```python
blank_counts = {
    column: df[column].astype(str).str.strip().eq("").sum()
    for column in df.select_dtypes(include="object").columns
}
{column: count for column, count in blank_counts.items() if count > 0}
```

```python
total_charges_numeric = pd.to_numeric(df["TotalCharges"], errors="coerce")
print("TotalCharges source dtype:", df["TotalCharges"].dtype)
print("Missing after numeric conversion:", total_charges_numeric.isna().sum())
df.loc[total_charges_numeric.isna(), ["customerID", "tenure", "TotalCharges", "Churn"]]
```

```python
print("Rows:", len(df))
print("Unique customer IDs:", df["customerID"].nunique())
for column in ["SeniorCitizen", "tenure", "MonthlyCharges"]:
    print(column, "min=", df[column].min(), "max=", df[column].max())
```

## Day 1 interpretation

This notebook intentionally stops at contract-oriented inspection. `customerID` is treated as a traceability identifier rather than a model feature. `TotalCharges` is inspected for source-format blanks and is normalized only by the validated loader. Deeper EDA, preprocessing fitting, model training, and evaluation belong to later days.
